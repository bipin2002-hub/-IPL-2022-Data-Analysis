```python
# Which team won the most mathes
```

```python
match_win=df["match_winner"].value_counts()
match_win.head(10)
```

```python
sns.barplot(y=match_win.index,x=match_win.values,palette="rainbow")
plt.show()
```

```python
# Toss decision Trend
toss_decision=df["toss_decision"].value_counts()
print(toss_decision)
sns.countplot(x="toss_decision",data=df)
plt.show()
```

```python
# Toss Winner Vs Match Winner
count=df[df["toss_winner"]== df["match_winner"]]["match_id"].count()
count
percentage=count/len(df["match_winner"])
print((percentage*100).round(2))
```

```python
#  How do team Win(run Vs Wickets)
won_by=df["won_by"].value_counts()
print(won_by)
sns.countplot(x=df["won_by"])
```

```python
### key Player Performance
```

```python
# most player of the match Award
count=df["player_of_the_match"].value_counts()
count.head(10)
```

```python
sns.barplot(x = count.head(10).values , y = count.head(10).index,palette='mako')
plt.title("Top 10 players with man of the match")
```

```python
#  Top Score
Top_score=df.groupby("top_scorer")["highscore"].sum().sort_values(ascending=False)
Top_score.head(10)
```

```python
Top_score.head(10).plot(kind="barh")
```

```python
# 10 best Bowler
df.head()
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowlers.plot(kind = 'barh')
```
